# Verifying Relationships Between Tables

```python

# All customer_id in orders must exist in customers.
print(orders['customer_id'].isin(customers['customer_id']).all())
# All order_id in order_items must exist in orders.
print(order_items['order_id'].isin(orders['order_id']).all())
# All order_id in order_payments must exist in orders.
print(order_payments['order_id'].isin(orders['order_id']).all())
# All order_id in order_reviews must exist in orders.
print(order_reviews['order_id'].isin(orders['order_id']).all())
# All product_id in order_items must exist in products.
print(order_items['product_id'].isin(products['product_id']).all())
# All seller_id in order_items must exist in sellers.
print(order_items['seller_id'].isin(sellers['seller_id']).all())
# All product_category_name in products must exist in category_translation.
products['product_category_name'].dropna().isin(category_translation['product_category_name']).all()
# All customer_zip_code_prefix in customers must exist in geolocation geolocation_zip_code_prefix.
print(customers['customer_zip_code_prefix'].isin(geolocation['geolocation_zip_code_prefix']).all())
# All seller_zip_code_prefix in sellers must exist in geolocation geolocation_zip_code_prefix.
print(sellers['seller_zip_code_prefix'].isin(geolocation['geolocation_zip_code_prefix']).all())
```

#  Strong EDA Analysis

```python
# Monthly Order Trend (Business Growth)
orders['order_purchase_timestamp']=pd.to_datetime(orders['order_purchase_timestamp'])
monthly_orders=orders.groupby(orders['order_purchase_timestamp'].dt.to_period('M')).size()
monthly_orders.plot(kind='line',figsize=(10,5),title="Monthly Order Trend")
```

```python
# Orders by State (Customer Distribution)
customer_orders=orders.merge(customers,on='customer_id')
customer_orders['customer_state'].value_counts().head(10).plot(kind='bar',title="Top 10 States by Number of Orders")
```

```python
# Average Order Value (AOV)
avg_order_value=order_payments.groupby('order_id')['payment_value'].sum().mean()
print("Average Order Value:", avg_order_value)
```

```python
# Products per Order (Basket Size)
items_per_order = order_items.groupby('order_id').size()
items_per_order.describe()
```

```python
items_per_order.value_counts().head(10).plot(kind='bar',title="Products per Order Distribution")
```
